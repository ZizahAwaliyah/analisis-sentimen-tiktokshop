```python
import pandas as pd
import glob
import os

# GANTI path di bawah ini jika perlu.
# Path yang Anda berikan sebelumnya: '/content/drive/MyDrive/SNA/CSV/'
path_folder_xlsx = '/content/drive/MyDrive/SEM 5/SNA/bu novy/'

# Membuat pola untuk menemukan semua file yang berakhiran .xlsx
pola_file = os.path.join(path_folder_xlsx, "*.xlsx")

# Mengambil daftar semua nama file xlsx
semua_file_xlsx = glob.glob(pola_file)

# Memeriksa apakah file ditemukan
if not semua_file_xlsx:
    print(f"Error: Tidak ada file .xlsx yang ditemukan di folder '{path_folder_xlsx}'")
    print("Mohon periksa kembali path folder dan pastikan ekstensinya sudah benar .xlsx")
else:
    # Membaca setiap file xlsx dan menyimpannya dalam sebuah list
    list_df = []
    print(f"Menemukan {len(semua_file_xlsx)} file. Memulai proses penggabungan...")
    for file in semua_file_xlsx:
        try:
            # Menggunakan pd.read_excel untuk file .xlsx
            df_temp = pd.read_excel(file)
            list_df.append(df_temp)
            print(f"-> Berhasil membaca {os.path.basename(file)}")
        except Exception as e:
            print(f"Gagal membaca file {file}: {e}")

    # Menggabungkan semua dataframe dalam list menjadi satu dataframe tunggal
    if list_df:
        df_gabungan = pd.concat(list_df, ignore_index=True)

        # Menampilkan informasi ringkas tentang dataframe gabungan
        print("\n================ INFORMASI DATAFRAME GABUNGAN ================")
        print(f"Berhasil menggabungkan {len(semua_file_xlsx)} file XLSX.")
        print(df_gabungan.info())

        # Menampilkan 5 baris pertama dari dataframe gabungan
        print("\n================ 5 BARIS PERTAMA DATA ================")
        print(df_gabungan.head())
    else:
        print("Tidak ada data untuk digabungkan.")
```

```python
# --- Tambahkan kode ini di akhir skrip Anda ---

# 1. Tentukan path LENGKAP di Google Drive untuk menyimpan file hasil
# Pastikan path ini benar dan Anda memiliki izin tulis
path_hasil_drive = '/content/drive/MyDrive/SEM 5/SNA/bu novy/HASIL_GABUNGAN.xlsx'

# 2. Simpan dataframe gabungan ke file Excel
# index=False agar nomor baris (indeks) pandas tidak ikut disimpan
print(f"\nMenyimpan hasil gabungan ke Google Drive di: {path_hasil_drive}...")
df_gabungan.to_excel(path_hasil_drive, index=False)

print("File berhasil disimpan ke Google Drive Anda.")
```
